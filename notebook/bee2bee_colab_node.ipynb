# Bee2Bee node on Google Colab

Serve a Hugging Face model from a free Colab GPU and join the Bee2Bee mesh.
The P2P port is exposed through an ngrok tunnel (a free ngrok account and auth token are required).

> Colab sessions are temporary. Your node leaves the mesh when the runtime stops.

```python
# @title 1. Install
!pip install -q "bee2bee[hf,torch]" pyngrok
```

```python
# @title 2. Configure
MODEL = "Qwen/Qwen2.5-0.5B-Instruct"  # @param {type:"string"}
REGION = "colab"  # @param {type:"string"}
REGISTRY_URL = "https://coithub.org"  # @param {type:"string"}

from google.colab import userdata
from pyngrok import ngrok

# Store your token under Colab 'Secrets' as NGROK_AUTHTOKEN (never paste it into the notebook).
ngrok.set_auth_token(userdata.get("NGROK_AUTHTOKEN"))
tunnel = ngrok.connect(4003, "http")
public = tunnel.public_url.replace("https://", "wss://")
print("Public P2P address:", public)
```

```python
# @title 3. Run the node
import os
os.environ.update({
    "BEE2BEE_ANNOUNCE_ADDR": public,
    "BEE2BEE_REGISTRY_URL": REGISTRY_URL,
    "BEE2BEE_UPNP": "false",
    "BEE2BEE_REGION": REGION,
})
!bee2bee serve-hf --model "$MODEL" --port 4003 --api-port 4002
```
